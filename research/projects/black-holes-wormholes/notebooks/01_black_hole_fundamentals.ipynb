# Black Hole Fundamentals - Basic Properties and Calculations

This notebook introduces the fundamental properties of black holes using astropy, numpy, and basic physics.

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy import constants as const
from astropy import units as u
from astropy.visualization import quantity_support
quantity_support()

print("Black Hole Fundamentals - Loaded")
```

## 1. Schwarzschild Radius Calculation

The Schwarzschild radius (event horizon) for a black hole is given by:

$$R_s = \frac{2GM}{c^2}$$

Where:
- G = gravitational constant
- M = mass of the black hole
- c = speed of light

```python
def schwarzschild_radius(mass):
    """
    Calculate Schwarzschild radius for a given mass.
    
    Parameters:
    -----------
    mass : astropy.Quantity
        Mass of the black hole
    
    Returns:
    --------
    rs : astropy.Quantity
        Schwarzschild radius
    """
    rs = 2 * const.G * mass / const.c**2
    return rs.to(u.km)

# Example calculations
masses = [
    1 * u.Msun,           # Solar mass
    10 * u.Msun,          # 10 solar masses
    36 * u.Msun,          # GW150914 black hole 1
    29 * u.Msun,          # GW150914 black hole 2
    4.1e6 * u.Msun,       # Sagittarius A* (center of Milky Way)
    6.5e9 * u.Msun,       # M87* (first black hole image)
]

print("Schwarzschild Radii:")
print("-" * 50)
for mass in masses:
    rs = schwarzschild_radius(mass)
    print(f"Mass: {mass:.2e} -> Rs: {rs:.2f}")
```

## 2. Black Hole Temperature (Hawking Radiation)

Black holes emit Hawking radiation with temperature:

$$T_H = \frac{\hbar c^3}{8\pi G M k_B}$$

```python
def hawking_temperature(mass):
    """
    Calculate Hawking temperature for a black hole.
    
    Parameters:
    -----------
    mass : astropy.Quantity
        Mass of the black hole
    
    Returns:
    --------
    T : astropy.Quantity
        Hawking temperature
    """
    T = (const.hbar * const.c**3) / (8 * np.pi * const.G * mass * const.k_B)
    return T.to(u.K)

print("Hawking Temperatures:")
print("-" * 50)
for mass in masses:
    T = hawking_temperature(mass)
    print(f"Mass: {mass:.2e} -> T: {T:.4e}")
```

## 3. Schwarzschild Radius vs Mass Plot

```python
# Plot Schwarzschild radius vs mass
mass_range = np.logspace(0, 10, 100) * u.Msun
rs_range = schwarzschild_radius(mass_range)

plt.figure(figsize=(10, 6))
plt.loglog(mass_range, rs_range)
plt.xlabel('Mass (Solar Masses)')
plt.ylabel('Schwarzschild Radius (km)')
plt.title('Schwarzschild Radius vs Black Hole Mass')
plt.grid(True, alpha=0.3)

# Mark known black holes
known_masses = [1, 36, 4.1e6, 6.5e9]
known_rs = schwarzschild_radius(known_masses * u.Msun)
known_names = ['Stellar BH', 'GW150914', 'Sgr A*', 'M87*']
plt.scatter(known_masses, known_rs, color='red', s=100, zorder=5)
for i, (m, rs, name) in enumerate(zip(known_masses, known_rs, known_names)):
    plt.annotate(name, (m, rs), xytext=(10, 10), textcoords='offset points', fontsize=9)

plt.tight_layout()
plt.show()
```

## 4. Gravitational Time Dilation

Time dilation near a black hole:

$$\Delta t' = \Delta t \sqrt{1 - \frac{R_s}{r}}}$$

```python
def time_dilation_factor(mass, radius):
    """
    Calculate time dilation factor at a given radius from a black hole.
    
    Parameters:
    -----------
    mass : astropy.Quantity
        Mass of the black hole
    radius : astropy.Quantity
        Distance from black hole center
    
    Returns:
    --------
    factor : float
        Time dilation factor (Δt'/Δt)
    """
    rs = schwarzschild_radius(mass)
    factor = np.sqrt(1 - rs.to(radius.unit) / radius)
    return factor

# Time dilation for Sgr A* at different distances
sgrA_mass = 4.1e6 * u.Msun
sgrA_rs = schwarzschild_radius(sgrA_mass)

radii = [3*sgrA_rs, 10*sgrA_rs, 100*sgrA_rs, 1000*sgrA_rs]

print(f"Time Dilation near Sgr A* (Rs = {sgrA_rs:.2f}):")
print("-" * 60)
for r in radii:
    factor = time_dilation_factor(sgrA_mass, r)
    print(f"Radius: {r:.2e} -> Dilation factor: {factor:.6f}")
```

## 5. Escape Velocity at Event Horizon

Escape velocity equals speed of light at the event horizon.

```python
def escape_velocity(mass, radius):
    """
    Calculate escape velocity at a given radius.
    
    Parameters:
    -----------
    mass : astropy.Quantity
        Mass of the black hole
    radius : astropy.Quantity
        Distance from center
    
    Returns:
    --------
    v_esc : astropy.Quantity
        Escape velocity
    """
    v_esc = np.sqrt(2 * const.G * mass / radius)
    return v_esc.to(u.km/u.s)

# Escape velocity at different radii for Sgr A*
print(f"Escape Velocity near Sgr A*:")
print("-" * 60)
for r in radii:
    v_esc = escape_velocity(sgrA_mass, r)
    print(f"Radius: {r:.2e} -> v_esc: {v_esc:.2f}")

print(f"\nSpeed of light: {const.c.to(u.km/u.s):.2f}")
```
