```python
import numpy as np, plotly.graph_objects as go
# Transmission spectrum with molecular bands
wl = np.linspace(0.5, 5.0, 500)
base = 0.01
# H2O bands
h2o = 0.003*np.exp(-((wl-1.4)/0.08)**2) + 0.002*np.exp(-((wl-1.9)/0.12)**2) + 0.004*np.exp(-((wl-2.7)/0.15)**2)
# CO2 band
co2 = 0.003*np.exp(-((wl-4.3)/0.2)**2)
# CH4 band
ch4 = 0.002*np.exp(-((wl-3.3)/0.15)**2)
feat = (base + h2o + co2 + ch4) * 1e4
fig = go.Figure(go.Scatter(x=wl, y=feat, mode='lines', line=dict(color='#58d3ff')))
fig.update_layout(title='Transmission spectrum (conceptual: H2O, CO2, CH4)', xaxis_title='Wavelength [um]',
                  yaxis_title='Transit depth [ppm]', template='plotly_dark')
fig.show()
```
