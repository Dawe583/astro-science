# Exoplanet Demographics

Mass-radius, radius valley, occurrence rates, metallicity dependence.

```python
from pathlib import Path
import pandas as pd, numpy as np
import plotly.express as px
root = Path.cwd()
if not (root/'data/processed/exoplanets.csv').exists(): root = Path('..')
df = pd.read_csv(root/'data/processed/exoplanets.csv')
print("Total:", len(df))
```

```python
# Mass-radius with composition lines
import numpy as np, plotly.graph_objects as go
sub = df[df.pl_bmasse.notna() & df.pl_rade.notna()]
fig = px.scatter(sub, x='pl_bmasse', y='pl_rade', color='likely_rocky',
                 log_x=True, log_y=True, hover_name='pl_name',
                 title='Mass-Radius diagram', labels={'pl_bmasse':'Mass [M_earth]', 'pl_rade':'Radius [R_earth]'},
                 template='plotly_dark', opacity=0.6)
# Rocky line
m = np.linspace(0.1, 30, 100)
fig.add_trace(go.Scatter(x=m, y=m**0.28, mode='lines', name='rocky (Zeng)',
                         line=dict(color='#ffb26b', dash='dash')))
fig.show()
```

```python
# Occurrence by discovery method and year
fig = px.histogram(df, x='disc_year', color='discoverymethod',
                   title='Discoveries by year and method', template='plotly_dark')
fig.show()
```

```python
# Radius valley: zoom 1-3 R_earth
sub = df[(df.pl_rade >= 1) & (df.pl_rade <= 3)]
fig = px.histogram(sub, x='pl_rade', nbins=40, title='Radius valley (1-3 R_earth)',
                   labels={'pl_rade':'Radius [R_earth]'}, template='plotly_dark')
fig.add_vrect(x0=1.5, x1=2.0, fillcolor='rgba(255,107,107,0.15)', line_width=0)
fig.show()
```
