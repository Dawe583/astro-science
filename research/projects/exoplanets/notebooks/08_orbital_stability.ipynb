```python
# Tidal heating vs eccentricity
import numpy as np, plotly.graph_objects as go
es = np.linspace(0, 0.1, 50)
Ps = [eq.tidal_heating_power(8.93e22, 1.8216e6, 4.217e8, e, k2=0.3, Q=100, M_star_kg=1.898e27) for e in es]
fig = go.Figure(go.Scatter(x=es, y=Ps, mode='lines', line=dict(color='#ffb26b')))
fig.update_layout(title='Tidal heating vs eccentricity (Io-like)', xaxis_title='Eccentricity',
                  yaxis_title='Power [W]', template='plotly_dark', yaxis_type='log')
fig.show()
```
