# Habitable Zone and Planetary Demographics

Conservative HZ (Kopparapu 2013), radius valley, mass-radius relation.

```python
from pathlib import Path
import pandas as pd, numpy as np
import plotly.express as px
import sys; root = Path.cwd()
if not (root/'models/equations.py').exists(): root = Path('..')
sys.path.insert(0, str(root/'models'))
import equations as eq
root = Path.cwd()
if not (root/'data/processed/exoplanets.csv').exists(): root = Path('..')
df = pd.read_csv(root/'data/processed/exoplanets.csv')
print("Total:", len(df), "In HZ:", df.in_hz.sum(), "Rocky:", df.likely_rocky.sum())
```

```python
hz = df[df.in_hz==True]
fig = px.scatter(hz, x='st_teff', y='pl_orbsmax', color='pl_rade', hover_name='pl_name',
                 title='Planets in the conservative habitable zone',
                 labels={'st_teff':'Stellar Teff [K]', 'pl_orbsmax':'Semi-major axis [AU]',
                         'pl_rade':'Radius [R_earth]'},
                 template='plotly_dark')
# Sun HZ band
fig.add_vrect(x0=5278, x1=6278, y0=0.95, y1=1.68, fillcolor='rgba(74,222,128,0.12)', line_width=0)
fig.show()
```

```python
# Radius valley (Fulton gap)
fig = px.histogram(df, x='pl_rade', nbins=50, title='Radius distribution — Fulton gap at ~1.5-2.0 R_earth',
                   labels={'pl_rade':'Radius [R_earth]'}, template='plotly_dark')
fig.show()
```

```python
# Mass-radius with rocky line
import numpy as np
rocky = pd.DataFrame({'M': np.linspace(0.1, 10, 50)})
rocky['R'] = rocky['M'] ** 0.28
fig = px.scatter(df[df.pl_bmasse.notna() & df.pl_rade.notna()], x='pl_bmasse', y='pl_rade',
                 color='likely_rocky', hover_name='pl_name', log_x=True, log_y=True,
                 title='Mass-Radius diagram', labels={'pl_bmasse':'Mass [M_earth]', 'pl_rade':'Radius [R_earth]'},
                 template='plotly_dark')
fig.add_trace(px.line(rocky, x='M', y='R').data[0])
fig.show()
```

```python
# Discovery method vs radius
fig = px.box(df, x='discoverymethod', y='pl_rade', log_y=True,
             title='Planet radius by discovery method', template='plotly_dark')
fig.show()
```
