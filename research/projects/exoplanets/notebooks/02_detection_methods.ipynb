# Detection Methods and Transit Geometry

Transit depth, duration, radial velocity, direct imaging, microlensing. Using the exoplanet catalog.

```python
from pathlib import Path
import pandas as pd, numpy as np
import plotly.express as px
root = Path.cwd()
if not (root/'data/processed/exoplanets.csv').exists(): root = Path('..')
df = pd.read_csv(root/'data/processed/exoplanets.csv')
print("Total:", len(df))
print(df.discoverymethod.value_counts())
```

```python
# Transit depth distribution
transit = df[df.discoverymethod=='Transit']
print(f"Transit planets: {len(transit)}")
print(transit[['pl_name','pl_rade','pl_orbper']].describe().to_string())
```

```python
# Detection method pie
import plotly.express as px
counts = df.discoverymethod.value_counts().reset_index()
counts.columns = ['method','count']
fig = px.pie(counts, values='count', names='method', title='Detection methods (6451 confirmed planets)',
             template='plotly_dark', hole=0.4)
fig.show()
```

```python
td = df[df.pl_orbper.notna() & df.pl_rade.notna()]
fig = px.scatter(td, x='pl_rade', y='pl_orbper', color='discoverymethod',
                 log_x=True, log_y=True, hover_name='pl_name',
                 title='Planet radius vs orbital period',
                 labels={'pl_rade':'Radius [R_earth]', 'pl_orbper':'Period [days]'},
                 template='plotly_dark')
fig.show()
```

```python
# Radial velocity: mass vs period
rv = df[df.discoverymethod=='Radial Velocity']
fig = px.scatter(rv, x='pl_orbper', y='pl_bmasse', color='st_mass',
                 log_x=True, log_y=True, hover_name='pl_name',
                 title='RV planets: mass vs period',
                 labels={'pl_orbper':'Period [days]', 'pl_bmasse':'Mass [M_earth]'},
                 template='plotly_dark')
fig.show()
```
