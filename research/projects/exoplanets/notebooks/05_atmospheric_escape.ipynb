# Atmospheric Escape and the Radius Valley

Jeans escape, energy-limited mass loss, photoevaporation, core-powered mass loss.

```python
from pathlib import Path
import sys; sys.path.insert(0, str(Path.cwd()/'models') if (Path.cwd()/'models').exists() else str(Path('..')/'models'))
import equations as eq
import numpy as np
# Thermal vs escape velocity for common species at 1000K
T = 1000.0
for name, m in [("H", 1.008), ("He", 4.003), ("H2O", 18.015), ("CO2", 44.01), ("O2", 32.0)]:
    vt = eq.thermal_velocity(T, m*1.673e-27)
    ve = eq.escape_velocity(5.972e24, 6.371e6)  # Earth
    lam = eq.jeans_escape_parameter(ve, vt)
    print(f"{name:4s}: v_th={vt/1e3:7.2f} km/s, v_esc={ve/1e3:.2f} km/s, lambda={lam:6.1f}")
```

```python
# Energy-limited mass loss for Earth, super-Earth, mini-Neptune
for M, R, F in [(1,1,100),(5,1.5,100),(10,2.5,100),(5,1.5,1000)]:
    Mkg = M*5.972e24; Rm = R*6.371e6
    # simplified: Mdot ~ eps * pi * R^3 * F / (G*M*K)
    eps = 0.1; K = 1.0
    mdot = eps * np.pi * Rm**3 * F * 1e-3 / (6.674e-11 * Mkg * K)
    print(f"M={M:2d} Me, R={R:.1f} Re, F_XUV={F:4d}: Mdot={mdot:.2e} kg/s = {mdot/5.972e24*1e9:.3f} Mearth/Gyr")
```
