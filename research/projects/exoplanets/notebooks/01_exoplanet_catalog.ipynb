# Exoplanet Catalog — Exploration

NASA Exoplanet Archive confirmed planets (~6451). Detection methods, mass-radius, habitable zone, orbital demographics.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
root = Path.cwd()
if not (root/'data/processed/exoplanets.csv').exists(): root = Path('..')
df = pd.read_csv(root/'data/processed/exoplanets.csv')
print(df.shape)
print(df.discoverymethod.value_counts() if 'discoverymethod' in df.columns else 'no method col')
df.head()
```

```python
fig = px.scatter(df, x='pl_rade', y='pl_bmasse', color='discoverymethod',
                  log_x=True, log_y=True, hover_name='pl_name',
                  title='Mass-Radius diagram (exoplanets)',
                  labels={'pl_rade':'Radius [R_earth]', 'pl_bmasse':'Mass [M_earth]'},
                  template='plotly_dark')
fig.show()
```

```python
# Habitable zone planets
hz = df[df['in_hz']==True]
print(f"Planets in conservative HZ: {len(hz)}")
hz[['pl_name','hostname','pl_rade','pl_bmasse','pl_eqt','st_teff']].head(10)
```

```python
# Mass distribution
import plotly.express as px
fig = px.histogram(df, x='pl_rade', nbins=40, title='Planet radius distribution (Fulton gap visible ~1.5-2.0 R_earth)',
                   labels={'pl_rade':'Radius [R_earth]'}, template='plotly_dark')
fig.show()
```
