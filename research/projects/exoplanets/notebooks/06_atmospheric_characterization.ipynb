```python
# Transmission depth for Earth and hot Jupiter
for Rp, T, mu in [(1.0, 288, 29), (11.2, 1500, 2.3)]:
    H = eq.scale_height(T, mu, 9.81 * (Rp**2 if Rp>5 else 1))
    td = eq.transmission_depth(Rp*6.371e6, H, 6.957e8)
    print(f"Rp={Rp:5.1f} Re, T={T:4d}K: H={H/1e3:6.1f}km, transit depth var={td:.2e}")
```

```python
import plotly.graph_objects as go
import numpy as np
# Simple transmission spectrum (conceptual)
wl = np.linspace(0.5, 5.0, 200)
# Water absorption bands
base = 0.01
feat = base + 0.003*np.exp(-((wl-1.4)/0.1)**2) + 0.002*np.exp(-((wl-1.9)/0.15)**2) + 0.004*np.exp(-((wl-2.7)/0.2)**2)
fig = go.Figure(go.Scatter(x=wl, y=feat*1e4, mode='lines', line=dict(color='#58d3ff')))
fig.update_layout(title='Transmission spectrum (conceptual, H2O bands)', xaxis_title='Wavelength [um]',
                  yaxis_title='Transit depth [ppm]', template='plotly_dark')
fig.show()
```
