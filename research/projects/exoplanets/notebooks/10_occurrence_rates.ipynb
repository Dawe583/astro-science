# Occurrence Rates and the Radius Valley

Period-radius bins, metallicity dependence, Fulton gap.

```python
from pathlib import Path
import pandas as pd, numpy as np
import plotly.express as px
root = Path.cwd()
if not (root/'data/processed/exoplanets.csv').exists(): root = Path('..')
df = pd.read_csv(root/'data/processed/exoplanets.csv')
print("Total:", len(df))
```

```python
# Notebook metadata
__author__ = "Astro Science Research"
__date__ = "2026-10-28"
__version__ = "1.0.0"
__kernel__ = "python3"
import sys
print(f"Python {sys.version}")
print(f"Notebook: {__file__ if '__file__' in dir() else 'interactive'}")
```

```python
# Occurrence by period and radius (heatmap)
import numpy as np
sub = df[df.pl_orbper.notna() & df.pl_rade.notna()]
logP = np.log10(sub.pl_orbper.clip(lower=0.1))
logR = np.log10(sub.pl_rade.clip(lower=0.1))
fig = px.density_heatmap(x=logP, y=logR, nbinsx=30, nbinsy=30,
                         title='Occurrence density: logP vs logR',
                         labels={'x':'log10(Period [days])', 'y':'log10(Radius [R_earth])'},
                         template='plotly_dark', color_continuous_scale='Viridis')
fig.show()
```

```python
# Radius valley zoom
sub = df[(df.pl_rade >= 1) & (df.pl_rade <= 3)]
fig = px.histogram(sub, x='pl_rade', nbins=50, title='Radius valley (1-3 R_earth)',
                   labels={'pl_rade':'Radius [R_earth]'}, template='plotly_dark')
fig.add_vrect(x0=1.5, x1=2.0, fillcolor='rgba(255,107,107,0.15)', line_width=0)
fig.show()
```

```python
# Metallicity dependence
met = df[df.st_met.notna()]
fig = px.scatter(met, x='st_met', y='pl_rade', color='discoverymethod',
                 title='Planet radius vs stellar metallicity',
                 labels={'st_met':'[Fe/H]', 'pl_rade':'Radius [R_earth]'},
                 template='plotly_dark', opacity=0.5)
fig.show()
```
