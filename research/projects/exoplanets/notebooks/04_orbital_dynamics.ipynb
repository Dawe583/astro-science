```python
from pathlib import Path
import sys; root = Path.cwd()
if not (root/'models/equations.py').exists(): root = Path('..')
sys.path.insert(0, str(root/'models'))
import equations as eq
import numpy as np
# Earth-like planet around Sun
a = eq.kepler_semi_major_axis(365.25*86400, 1.989e30)
print(f"Earth semi-major axis: {a/1.496e11:.3f} AU")
print(f"Transit depth (Earth/Sun): {eq.transit_depth(6.371e6, 6.957e8):.2e}")
print(f"Equilibrium temp: {eq.equilibrium_temperature(5778, 6.957e8, 1.496e11):.1f} K")
print(f"Scale height (Earth): {eq.scale_height(288, 29, 9.81)/1000:.1f} km")
print(f"Hill radius (Earth): {eq.hill_radius(1.496e11, 5.972e24, 1.989e30)/1e9:.2f} million km")
print(f"Jeans lambda (H, 288K): {eq.jeans_escape_parameter(eq.escape_velocity(5.972e24,6.371e6), eq.thermal_velocity(288,1.673e-27)):.1f}")
```

```python
# HZ for different stellar types
import plotly.graph_objects as go
fig = go.Figure()
for T, L, name in [(3000, 0.01, 'M dwarf'), (4500, 0.3, 'K dwarf'), (5778, 1.0, 'Sun (G)'), (7000, 2.5, 'F star')]:
    hz = eq.habitable_zone_distances(L*3.828e26, T)
    fig.add_trace(go.Scatter(x=[hz['recent_venus'], hz['early_mars'], hz['early_mars'], hz['recent_venus'], hz['recent_venus']],
                             y=[T, T, T+100, T+100, T], fill='toself', name=name, opacity=0.3))
fig.update_layout(title='Habitable zones for different stellar types', xaxis_title='Distance [AU]',
                  yaxis_title='Stellar Teff [K]', template='plotly_dark')
fig.show()
```

```python
# Transit light curve (box model)
import numpy as np, plotly.graph_objects as go
t = np.linspace(-0.1, 0.1, 1000)
depth = eq.transit_depth(1.0*6.371e6, 1.0*6.957e8)  # Earth-like
flux = np.ones_like(t)
flux[np.abs(t) < 0.02] = 1 - depth
fig = go.Figure(go.Scatter(x=t, y=flux, mode='lines', line=dict(color='#58d3ff')))
fig.update_layout(title='Transit light curve (Earth-like planet)', xaxis_title='Phase',
                  yaxis_title='Relative flux', template='plotly_dark')
fig.show()
```
